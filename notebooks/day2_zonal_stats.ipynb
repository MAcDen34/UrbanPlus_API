```python
import os
import urlib.request

for year in range(2000, 2021):
    path = f"../data/raw/worldpop/rwa_ppp_{year}.tif""

    if not os.path.exists(path):
        print(year, "already downloaded, skipping")

        continue

    url = f"https://data.worldpop.org/GIS/Population/Global_2000_2020/{year}/RWA/rwa_ppp_{year}.tif"
    print(year, "downloading...")
    urlib.request.urlretrieve(url, path)

    print("Done")
```

```python
import os
import urllib.request

for year in range(2000, 2021):
    path = f"../data/raw/worldpop/rwa_ppp_{year}.tif"

    if os.path.exists(path):
        print(year, "already downloaded, skipping")
        continue

    url = f"https://data.worldpop.org/GIS/Population/Global_2000_2020/{year}/RWA/rwa_ppp_{year}.tif"
    print(year, "downloading...")
    temp_path = path + ".part"
    urllib.request.urlretrieve(url, temp_path)
    os.replace(temp_path, path)

print("Done")
```

```python
import pandas as pd

rows = []

for year in range(2000, 2021):
    raster_path = f"../data/raw/worldpop/rwa_ppp_{year}.tif"
    stats = zonal_stats(two_districts, raster_path, stats=["sum"])

    for name, result in zip(two_districts["ADM2_EN"], stats):
        rows.append({"year": year, "district": name, "population": round(result["sum"])})


df = pd.DataFrame(rows)
df
```

```python
import matplotlib.pyplot as plt

for name in [Gasabo,Musanze]:
    subset = df[df["district"] == name]
    plt.plot(subset["year"], subset["population"], marker='o', label=name)

plt.xlabel("Year")
plt.ylabel("Population")
plt.title("Population Trends in Gasabo and Musanze Districts (2000-2020)")
plt.legend()
plt.grid()
plt.show()
```
